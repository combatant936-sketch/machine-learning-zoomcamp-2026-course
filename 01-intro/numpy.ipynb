```python
v.shape
```

```python
u+v
```

```python
def calculate_vector_vector_multiplication(a,b):
    assert a.shape[0]==b.shape[0]
    result=0.0
    shape=a.shape[0]

    for i in range(shape):
        result+= a[i]*b[i]

    return result
```

```python
calculate_vector_vector_multiplication(u,v)
```

```python
u.dot(v)
```

```python
U = np.array([
  [2, 4, 5, 6],
  [1, 2, 1, 2],
 [3, 1, 2, 1],
])
```

```python
U.shape
```

```python
v1=np.array([1,2,3,4])
```

```python
def calculate_matrix_vector_multiplication(U,v):
    assert U.shape[1]==v.shape[0]
    num_rows=U.shape[0]
    result=np.zeros(num_rows)

    for i in range(num_rows):
        result[i]=calculate_vector_vector_multiplication(U[i],v)

    return result
    
```

```python
calculate_matrix_vector_multiplication(U,v1)
```

```python
U.dot(v1)
```

```python
V = np.array([
[1, 1, 2],
[0, 0.5, 1], 
[0, 2, 1],
[2, 1, 0],
])
```

```python
U.shape[1]
```

```python
V.shape[0]
```

```python
def matrix_matrix_multiplication(U,V):
    assert U.shape[1]==V.shape[0]

    num_rows=U.shape[0]
    num_cols=V.shape[1]

    result=np.zeros((num_rows,num_cols))

    for i in range(num_cols):
        vi=V[:,i]

        result[:,i]=calculate_matrix_vector_multiplication(U,vi)
    return result

    
```

```python
matrix_matrix_multiplication(U,V)
```

```python
U.dot(V)
```
