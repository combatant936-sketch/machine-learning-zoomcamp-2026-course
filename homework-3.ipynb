```python
def make_model(features):
        dv = DictVectorizer(sparse=False)
        
        train_dict = df_train[features].to_dict(orient='records')
        val_dict = df_val[features].to_dict(orient='records')
        
    
        X_train = dv.fit_transform(train_dict)
    
        X_val = dv.transform(val_dict)
    
        model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
        # solver='lbfgs' is the default solver in newer version of sklearn
        # for older versions, you need to specify it explicitly
        model.fit(X_train, y_train)

        return model,X_val


def diff_og_without_feature(features):

    model,X_val=make_model(features)

    y_pred = model.predict_proba(X_val)[:, 1]

    coverted_decision = (y_pred >= 0.5)
    og= (y_val == coverted_decision).mean()

    for f_remove in features:
        r = calulate_without_features(f_remove,features)
        print(f" difference between og_score {og} and without_feature_{f_remove} is {og-r}") 

def calulate_without_features(f_remove,features):
   
    test_features=copy(features)
    test_features.remove(f_remove)
    model, X_val = make_model(test_features)

    y_pred = model.predict_proba(X_val)[:, 1]

    coverted_decision = (y_pred >= 0.5)
    r= (y_val == coverted_decision).mean()

    return r
    
       


        
       
        




        
```
