```python
X=df_first_seven.values
```

```python
XT=X.T
```

```python
import numpy as np
```

```python
def calculate_vector_vector_multiplication(a,b):
    assert a.shape[0]==b.shape[0]
    result=0.0
    shape=a.shape[0]

    for i in range(shape):
        result+= a[i]*b[i]

    return result

def calculate_matrix_vector_multiplication(U,v):
    assert U.shape[1]==v.shape[0]
    num_rows=U.shape[0]
    result=np.zeros(num_rows)

    for i in range(num_rows):
        result[i]=calculate_vector_vector_multiplication(U[i],v)

    return result


def matrix_matrix_multiplication(U,V):
    assert U.shape[1]==V.shape[0]

    num_rows=U.shape[0]
    num_cols=V.shape[1]

    result=np.zeros((num_rows,num_cols))

    for i in range(num_cols):
        vi=V[:,i]

        result[:,i]=calculate_matrix_vector_multiplication(U,vi)
    return result

    
```
